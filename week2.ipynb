# Week 2: FIND-S Algorithm
Finding the most specific hypothesis from training examples

```python
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

print(data)
```

```python
import pandas as pd

data = {
    'Sky': ["Sunny", "Sunny", "Rainy", "Sunny"],
    'AirTemp': ["Warm", "Warm", "Cold", "Warm"],
    'Humidity': ["Normal", "High", "High", "High"],
    'Wind': ["Strong", "Strong", "Strong", "Strong"],
    'Water': ["Warm", "Warm", "Warm", "Cool"],
    'Forecast': ["Same", "Same", "Change", "Change"],
    'Enjoysport': ["Yes", "Yes", "No", "Yes"],
}

df = pd.DataFrame(data)

hypothesis = None
print("Initial Hypothesis: None")

for i in range(len(df)):
    example = list(df.iloc[i, :-1])
    label = df.loc[i, 'Enjoysport']

    print("Step", i + 1)
    print("Example:", example)
    print("Enjoysport:", label)

    if label == "Yes":
        if hypothesis is None:
            hypothesis = example.copy()
            print("First positive example.")
            print("Initialize hypothesis.")
        else:
            for j in range(len(hypothesis)):
                if hypothesis[j] != example[j]:
                    hypothesis[j] = "?"
            print("Generalizing hypothesis.")
    else:
        print("Negative example.")
        print("No change to hypothesis.")

    print("Current Hypothesis:")
    print(hypothesis)

print("FINAL HYPOTHESIS")
print(hypothesis)
```
